```python
import pandas as pd

# Direct raw URL to the CSV dataset
url = "https://raw.githubusercontent.com/ArchanaInsights/Datasets/main/marketing_campaign.csv"

# Load dataset into pandas DataFrame
df = pd.read_csv(url)

# Display the first few rows to verify loading
print(df.head(10))
```

```python
#Exploratory Data Analysis (EDA) and Visualization

#Campaign Performance
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Set figure size and plot style
plt.figure(figsize=(8, 6))
sns.set_theme(style="whitegrid")

# Create scatter plot
sns.scatterplot(
    data=df, 
    x='Acquisition_Cost', 
    y='ROI', 
    alpha=0.7, 
    edgecolor='w', 
    s=70
)

# Set titles and labels
plt.title('Relationship between Acquisition Cost and ROI', fontsize=14, pad=12)
plt.xlabel('Acquisition Cost', fontsize=12)
plt.ylabel('ROI (Return on Investment)', fontsize=12)

plt.tight_layout()
plt.show()
```

```python
# Set figure size and plot style
plt.figure(figsize=(10, 6))
sns.set_theme(style="whitegrid")

# Create bar plot showing average Conversion_Rate by Channels_Used grouped by Campaign_Type
sns.barplot(
    data=df,
    x='Channel_Used',
    y='Conversion_Rate',
    hue='Campaign_Type',
    ci=None  # Displays average value (height of bars) without error bars
)

# Set labels and title
plt.title('Average Conversion Rate by Channels Used (Categorized by Campaign Type)', fontsize=14, pad=12)
plt.xlabel('Channel_Used', fontsize=12)
plt.ylabel('Average Conversion Rate', fontsize=12)
plt.legend(title='Campaign Type', bbox_to_anchor=(1.05, 1), loc='upper left')

plt.tight_layout()
plt.show()
```

```python
# Set figure size and plot style
plt.figure(figsize=(10, 6))
sns.set_theme(style="whitegrid")

# Create box plot showing distribution of Engagement_Score across Campaign_Type
sns.boxplot(
    data=df,
    x='Campaign_Type',
    y='Engagement_Score',
    palette='Set2'
)

# Set titles and labels
plt.title('Distribution of Engagement Score across Campaign Types', fontsize=14, pad=12)
plt.xlabel('Campaign Type', fontsize=12)
plt.ylabel('Engagement Score', fontsize=12)

plt.tight_layout()
plt.show()
```

```python
# Calculate average ROI by Company and sort for clearer comparison
avg_roi_company = df.groupby('Company')['ROI'].mean().reset_index().sort_values(by='ROI', ascending=False)

# Set figure size and plot style
plt.figure(figsize=(10, 6))
sns.set_theme(style="whitegrid")

# Create bar chart
sns.barplot(
    data=avg_roi_company,
    x='Company',
    y='ROI',
    palette='Blues_r'
)

# Set titles and labels
plt.title('Average ROI by Company', fontsize=14, pad=12)
plt.xlabel('Company', fontsize=12)
plt.ylabel('Average ROI', fontsize=12)
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
# Calculate correlation matrix between Engagement_Score and Conversion_Rate
corr = df[['Engagement_Score', 'Conversion_Rate']].corr()

# Set figure size and plot style
plt.figure(figsize=(6, 5))
sns.set_theme(style="white")

# Create heatmap
sns.heatmap(
    corr, 
    annot=True, 
    cmap='coolwarm', 
    fmt='.2f', 
    vmin=-1, 
    vmax=1, 
    cbar=True, 
    square=True,
    linewidths=0.5
)

# Set title
plt.title('Correlation between Engagement Score and Conversion Rate', fontsize=12, pad=12)

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

# Convert Date using the correct day-first format
df['Date'] = pd.to_datetime(df['Date'], format='%d-%m-%Y')

# Aggregate average Conversion_Rate by Date and Company
daily_company_conv = (
    df.groupby(['Date', 'Company'])['Conversion_Rate'].mean().reset_index()
)

# Set figure size and plot style
plt.figure(figsize=(12, 6))
sns.set_theme(style='whitegrid')

# Create line plot
sns.lineplot(
    data=daily_company_conv,
    x='Date',
    y='Conversion_Rate',
    hue='Company',
    marker='o',
)

# Title and labels
plt.title('Conversion Rate Trends Over Time by Company', fontsize=14, pad=12)
plt.xlabel('Date', fontsize=12)
plt.ylabel('Average Conversion Rate', fontsize=12)
plt.xticks(rotation=45)
plt.legend(title='Company', bbox_to_anchor=(1.05, 1), loc='upper left')

plt.tight_layout()
plt.show()
```

```python
# Convert Date to datetime format using Day-Month-Year parsing
df['Date'] = pd.to_datetime(df['Date'], format='%d-%m-%Y')

# Aggregate mean Engagement_Score by Date
daily_engagement = (
    df.groupby('Date')['Engagement_Score'].mean().reset_index()
)

# Set figure size and plot style
plt.figure(figsize=(12, 6))
sns.set_theme(style="whitegrid")

# Create line chart
sns.lineplot(
    data=daily_engagement,
    x='Date',
    y='Engagement_Score',
    color='teal',
    linewidth=2,
    marker='o'
)

# Set titles, labels, and tick formatting
plt.title('Trend of Average Engagement Score Over Time', fontsize=14, pad=12)
plt.xlabel('Date', fontsize=12)
plt.ylabel('Average Engagement Score', fontsize=12)
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
#Geographic Analysis
```

```python
# Calculate average Acquisition_Cost by Location and sort descending
avg_cost_location = (
    df.groupby('Location')['Acquisition_Cost']
    .mean()
    .reset_index()
    .sort_values(by='Acquisition_Cost', ascending=False)
)

# Set figure size and plot style
plt.figure(figsize=(10, 6))
sns.set_theme(style="whitegrid")

# Create bar chart
sns.barplot(
    data=avg_cost_location,
    x='Location',
    y='Acquisition_Cost',
    palette='Reds_d'
)

# Set title and labels
plt.title('Average Acquisition Cost by Location', fontsize=14, pad=12)
plt.xlabel('Location', fontsize=12)
plt.ylabel('Average Acquisition Cost', fontsize=12)
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
# Set a clean visual style
sns.set_theme(style="whitegrid")

# Set up a spacious figure size to fit all groups comfortably
plt.figure(figsize=(14, 8))

# Create the grouped bar chart
sns.barplot(
    data=df, 
    x='Location', 
    y='Conversion_Rate', 
    hue='Target_Audience', 
    estimator='mean',   # Focuses on the average conversion rate
    errorbar=None,       # Disables error bars for visual clarity
    palette='Set2'       # Distinct, professional color palette
)

# Customize titles and labels
plt.title('Average Conversion Rate by Location and Target Audience', fontsize=14, pad=15)
plt.xlabel('Location', fontsize=12)
plt.ylabel('Average Conversion Rate', fontsize=12)

# Clean up x-axis label formatting
plt.xticks(rotation=30, ha='right')

# Place the legend safely outside the plot matrix boundary line
plt.legend(title='Target Audience', bbox_to_anchor=(1.05, 1), loc='upper left')

# Adjust layout to prevent text clipping
plt.tight_layout()
plt.show()
```
